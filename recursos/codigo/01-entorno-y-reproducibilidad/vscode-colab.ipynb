# Conexión de VS Code con Google Colab

> La conexión se realiza por medio de la extención Colab dentro de VS Code.
> Colab ejecuta las instruciones del notebook dentro de una instancia en sus servidores.

Para crear una instancia, debes seguir estos pasos:
1. En el menú que aparece al presionar "Select Kernel" de VS Code, selecciona "Select Another Kernel.."

2. En el menú para elegir la fuente del kernel, debes seleccionar "Colab".

3. Para tener mayor control con los recursos que solicitas a Colab, selecciona "+ New Colab Server".

4. Selecciona si necesitas una instancia de sólo CPU o requieres uno con GPU o TPU (Considera que instancias con GPU y TPU tienen un tiempo de ejecución muy limitado).

5. Elige las características de tu instancia en el menú.

6. Nombra tu instancia.

7. Selecciona el kernel de Python.

8. Espera a que aparezca la descripción del kernel seleccionado en el botón para seleccionar kernel (antes tenía la etiqueta "Select Kernel").

9. Puedes seleccionar el kernel que tienes asignado de Colab para otros notebooks dentro de VS Code.


```python
import sys # biblioteca para manipular argumentos de linea de comandos

IN_COLAB = "google.colab" in sys.modules

if IN_COLAB:
	print("Ejecutando en Google Colab")
else:
	print("Ejecutando en un entorno local")

print(f'Version de Python: {sys.version}')
```

# Desconexión de la instancia de Google Colab

> Recuerda siempre desconectar las instancias de Google Colab cuando no las estés utilizando para evitar que corra tiempo de ejecución.
> Algunas instancias se desconectan y cierran dado un determinado tiempo de inactividad.

- Para desconectarte selecciona el boton de "Colab" en el menú del notebook, presiona "Remove server" y selecciona la instancia que quieres desconectar.

- También, puedes verificar ye eliminar las instancias desde la web de colab usando el botón "Colab" del menú del notebook y seleccionando "Open Colab Web". Luego, debes abrir un notebook y seleccionar en el menú Runtime > Manage sessions.
