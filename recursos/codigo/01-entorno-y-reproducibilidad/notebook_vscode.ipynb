## 👀 Se debe tener cuidado con el orden de ejecución de las celdas.

Observa el siguiente ejemplo:

```python
# Celda de código A
x = 1
```

```python
# Celda de código B
x = 7
```

```python
# Celda de código C
print(4 + x) # Imprimir el valor de y
```

Si ejecutas primero C, mostrará que no se ha definido x.

## 🧰 Botones útiles en VS Code

Durante el trabajo con notebooks en VS Code encontrarás varios controles que
permiten administrar la ejecución y el estado del notebook.

| Botón | ¿Para qué sirve? |
|---|---|
| **▶ Run All** | Ejecuta **todas las celdas** del notebook en orden, desde la primera hasta la última. |
| **↻ Restart** | **Reinicia el kernel**. Borra de la memoria las variables, funciones y objetos creados durante la ejecución, pero **no elimina el código ni las salidas visibles** de las celdas. |
| **Clear All Outputs** | **Elimina las salidas visibles** de todas las celdas. No reinicia el kernel ni borra las variables almacenadas en memoria. |
| **Jupyter Variables** | Permite consultar las **variables que existen actualmente en la memoria del kernel**, junto con información como su tipo y valor. |
| **Select Kernel** | Permite elegir el **kernel con el que se ejecutará el notebook**. En este taller, debe corresponder al entorno de Python que queremos utilizar. |


> [!IMPORTANT]
> **Reiniciar el kernel** y **borrar las salidas** no son lo mismo:
>
> - **Restart** → limpia la **memoria** del kernel.
> - **Clear All Outputs** → limpia lo que **vemos** en el notebook.
