## Preparación

- Colab: Conecta un ambiente python en Colab, establece la URL HTTPS pública del repositorio y ejecuta la celda. Si ya existe, se actualizará con `git pull`.

- Local: En VS Code, puedes usar la carpeta del repositorio local y un entorno con el paquete ipywidgets.

```python
from pathlib import Path
import sys
import subprocess

REPO_URL = "https://github.com/bettypau-open-knowledge/del-dataset-a-unet_reto-1.git"  # Sustituir por la URL real
EN_COLAB = Path("/content").exists() and "google.colab" in sys.modules

if EN_COLAB:
    PROYECTO = Path("/content/reto_unidad_1")
    if (PROYECTO / ".git").exists():
        subprocess.run(["git", "-C", str(PROYECTO), "pull", "--ff-only"], check=True)
    elif PROYECTO.exists():
        raise RuntimeError(f"Existe {PROYECTO}, pero no es un repositorio Git.")
    else:
        if "USUARIO" in REPO_URL:
            raise ValueError("Sustituye REPO_URL por la URL HTTPS pública del repositorio.")
        subprocess.run(["git", "clone", REPO_URL, str(PROYECTO)], check=True)
else:
    PROYECTO = Path.cwd()
    if not (PROYECTO / "src").is_dir():
        PROYECTO = PROYECTO.parent

if str(PROYECTO) not in sys.path:
    sys.path.insert(0, str(PROYECTO))
print("Proyecto:", PROYECTO)
```
